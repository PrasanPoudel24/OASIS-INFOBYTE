```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Unemployment in India.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df[' Date'] = pd.to_datetime(df[' Date'], errors='coerce')
```

```python
df.describe()
```

```python
# Convert to datetime if not already
df[' Date'] = pd.to_datetime(df[' Date'])

# Extract components
df['day'] = df[' Date'].dt.day
df['month'] = df[' Date'].dt.month
df['year'] = df[' Date'].dt.year
```

```python
df.columns
```

```python
df.head()
```

```python
df=df.drop(columns=[' Date'])
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df=df.dropna()
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['day']=df['day'].astype(int)
df['month']=df['month'].astype(int)
df['year']=df['year'].astype(int)
```

```python
df.info()
```

```python
df['year'].value_counts()
```

```python
df.duplicated().sum()
```

```python
df['Region'].value_counts()
```

```python
new=df.groupby('Region')[' Estimated Unemployment Rate (%)'].mean().reset_index()
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=new,
    x='Region',
    y=' Estimated Unemployment Rate (%)'
)
plt.xticks(rotation=90)
plt.title('Average Unemployment Rate by Region')
plt.xlabel('Region')
plt.ylabel('Average Unemployment Rate (%)')
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=new_two,
    x='Region',
    y=' Estimated Employed'
)

plt.xticks(rotation=90)
plt.title('Average Employment by Region')
plt.xlabel('Region')
plt.ylabel('Average Estimated Employed')
plt.show()
```

```python
new_three= df.groupby('Region')[' Estimated Labour Participation Rate (%)'].mean().reset_index()
```

```python
df.rename(columns={
    ' Frequency': 'Frequency',
    ' Estimated Unemployment Rate (%)': 'Unemployment Rate (%)',
    ' Estimated Employed': 'Employed',
    ' Estimated Labour Participation Rate (%)': 'Labour Participation Rate (%)'
}, inplace=True)
```

```python
df.columns
```

```python
month_names = {
    1: 'January',
    2: 'February',
    3: 'March',
    4: 'April',
    5: 'May',
    6: 'June',
    7: 'July',
    8: 'August',
    9: 'September',
    10: 'October',
    11: 'November',
    12: 'December'
}

df['month'] = df['month'].map(month_names)
```

```python
new=df.groupby('month')['Labour Participation Rate (%)'].mean().reset_index()
```

```python
plt.figure(figsize=(9, 9))

plt.pie(
    new['Labour Participation Rate (%)'],
    labels=new['month'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Average Labour Participation Rate by Month')
plt.show()
```
